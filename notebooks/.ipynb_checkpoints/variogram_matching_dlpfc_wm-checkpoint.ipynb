# Maynard DLPFC 151673 white matter: variance-constrained variogram matching

Visium sample **151673**, all spots annotated **WM** in `data/h5ad/151673.h5ad` (513 spots, all in tissue). This is the full annotated white-matter region, not a spatial subset. The ten prespecified genes are RSRP1, LRRC41, SLC17A7, SNAP25, CAMK2A, MBP, GFAP, NEFL, PLP1, and MOBP. Eight are from the Layer 3 diagnostic panel. CUX2 and RORB are excluded because they are detected in 10 and 20 of 513 white-matter spots. PLP1 and MOBP replace them. This is a targeted diagnostic panel, not a random gene sample.

Expression is natural-log CPM: \(\log(1+10^6 C_{ig}/L_i)\). \(L_i\) is the spot's total UMI count across all 33,538 genes stored in `X`; this file has no separate library-size column. Each gene is then standardized to population variance one. Coordinates retain their native spatial units. Pair assignments use the same exact 25th-percentile distance cutoff and 13 bins as the MERFISH notebook.

We fit \(a+b\gamma_Y(h)\) subject to \(a+b\operatorname{Var}(Y)=1\), \(a,b\geq0\), choosing bandwidth by constrained theoretical SSE. Surrogates are \(\sqrt bY+\sqrt aR\), with no final rescaling. Three seeds (42, 43, 44) use a **shared row permutation** and correlated noise \(R=WU/\sqrt N\), as in production, then center/standardize each noise column as in the constrained notebook. This latter operation does not preserve the noise covariance exactly. All candidates within a seed share the same permutation and noise.

Uniform bandwidth parameters are \([1,2,3,4,6,8,12,16,24,32,48,64]d\), where \(d\) is median nearest-neighbor distance and Gaussian SD is bandwidth/2.5. Compare the old grid through 12 with the expanded grid on **identical random draws**. Adaptive candidates reuse the MERFISH grid. The realized-variance and post-normalization diagnostics retain the sample cross term. This assesses fitting behavior, not downstream p-value calibration.

```python
from pathlib import Path
import h5py
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from anndata.io import read_elem, sparse_dataset
from scipy.spatial import cKDTree
from scipy.spatial.distance import pdist, cdist
from threadpoolctl import threadpool_limits
from IPython.display import display

root = next(p for p in (Path.cwd(), *Path.cwd().parents)
            if (p / "data/h5ad/151673.h5ad").is_file())
output_dir = root / "results/variogram_matching_maynard_dlpfc_wm"
output_dir.mkdir(parents=True, exist_ok=True)
selected_genes = np.array(["RSRP1", "LRRC41", "SLC17A7", "SNAP25", "CAMK2A",
                           "MBP", "GFAP", "NEFL", "PLP1", "MOBP"])
N_GENES = len(selected_genes)
SEED = 42
SEEDS = [42, 43, 44]
N_BINS = 13
with h5py.File(root / "data/h5ad/151673.h5ad", "r") as f:
    var = read_elem(f["var"])
    names = var.index.astype(str).to_numpy()
    obs = read_elem(f["obs"])
    region = obs["Region"].astype(str).to_numpy()
    layer_mask = region == "WM"
    selected_columns = []
    for name in selected_genes:
        indices = np.flatnonzero(names == name)
        if len(indices) != 1:
            raise ValueError(f"Expected one gene ID for {name}, got {len(indices)}")
        selected_columns.append(int(indices[0]))
    matrix = sparse_dataset(f["X"])
    # Read selected columns and the all-gene library size; do not sum only the panel.
    counts = np.asarray(matrix[:, selected_columns].tocsr()[layer_mask].toarray(), dtype=np.float64)
    library_size = np.asarray(matrix[layer_mask].sum(axis=1), dtype=np.float64).ravel()
    coords = np.asarray(read_elem(f["obsm/spatial"]), dtype=np.float64)[layer_mask]
    spot_ids = obs.index.astype(str).to_numpy()[layer_mask]
keep = (library_size > 0) & np.isfinite(coords).all(axis=1)
counts, coords, library_size, spot_ids = counts[keep], coords[keep], library_size[keep], spot_ids[keep]
assert np.isfinite(counts).all() and np.all(counts >= 0)
assert np.allclose(counts, np.round(counts))
assert np.all(counts.sum(axis=1) <= library_size + 1e-6)
log_cpm = np.log1p(1e6 * counts / library_size[:, None])
n_cells, n_genes = log_cpm.shape
vc_mean, vc_std = log_cpm.mean(axis=0), log_cpm.std(axis=0, ddof=0)
if np.any(vc_std <= 0):
    raise ValueError("Selected gene has zero variance.")
vc_true = (log_cpm - vc_mean) / vc_std
np.testing.assert_allclose(vc_true.var(axis=0), 1.0, atol=1e-12)
np.testing.assert_allclose(vc_true.mean(axis=0), 0.0, atol=1e-12)
# The geometry cell's baseline uses the same seed-42 shared permutation.
permuted_log_cpm = log_cpm[np.random.default_rng(SEED).permutation(n_cells)]
print(f"Maynard 151673, WM: {n_cells} spots; {n_genes} prespecified genes.")
display(pd.DataFrame({"gene": selected_genes, "detected_spots": (counts > 0).sum(axis=0),
                      "total_counts": counts.sum(axis=0), "log_CPM_variance": vc_std**2,
                      "standardized_variance": vc_true.var(axis=0)}))
fig, ax = plt.subplots(figsize=(6, 6))
ax.scatter(coords[:, 0], coords[:, 1], s=8)
ax.set(title="Maynard 151673: all annotated WM spots", xlabel="Spatial x (native units)", ylabel="Spatial y (native units)")
ax.set_aspect("equal")
plt.show()
```

```python
# Compute the exact cutoff from all unordered pairs, without diagonal zeros.
all_distances = pdist(coords, metric="euclidean")
max_dist = float(np.quantile(all_distances, 0.25))
total_pairs = all_distances.size
# geoR constructs bins up to the largest observed distance below max.dist.
below_cutoff = all_distances[all_distances < max_dist]
if below_cutoff.size == 0 or all_distances.min() <= 1e-12:
    raise ValueError("Expected distinct, separated coordinates.")
bin_max_dist = float(below_cutoff.max())
del all_distances, below_cutoff
if max_dist <= 0:
    raise ValueError("The 25th-percentile distance must be positive.")

# geoR 1.9-5 .define.bins: 13 equal-width intervals, 14 edges.
# Its separate near-zero bin is empty here and is dropped.
bin_edges = np.linspace(0.0, bin_max_dist, N_BINS + 1)
bin_edges[0] = 1e-12
lag_centers = (bin_edges[:-1] + bin_edges[1:]) / 2

# Only materialize pairs used by the variogram; each unordered pair occurs once.
pairs = cKDTree(coords).query_pairs(max_dist, output_type="ndarray")
pair_distances = np.linalg.norm(coords[pairs[:, 0]] - coords[pairs[:, 1]], axis=1)
# geoR binit uses left-closed, right-open bins, excluding the last edge.
inside = (pair_distances >= bin_edges[0]) & (pair_distances < bin_edges[-1])
pairs, pair_distances = pairs[inside], pair_distances[inside]
# Exact interior-edge distances belong to the bin on their right, as in geoR.
bin_index = np.searchsorted(bin_edges[1:-1], pair_distances, side="right")
pair_counts = np.bincount(bin_index, minlength=N_BINS)
valid_bins = pair_counts >= 2

def classical_variogram(expression):
    """Half the mean squared expression difference in each shared distance bin."""
    result = np.full((N_BINS, expression.shape[1]), np.nan)
    for j in range(expression.shape[1]):
        difference = expression[pairs[:, 0], j] - expression[pairs[:, 1], j]
        squared_sum = np.bincount(
            bin_index, weights=difference**2, minlength=N_BINS
        )
        result[valid_bins, j] = (
            squared_sum[valid_bins] / (2.0 * pair_counts[valid_bins])
        )
    return result

variogram_real = classical_variogram(log_cpm)
variogram_permuted = classical_variogram(permuted_log_cpm)
assert pair_counts.sum() == len(pairs)
assert np.all(variogram_real[valid_bins] >= 0)
assert np.all(variogram_permuted[valid_bins] >= 0)


vc_target_gamma = variogram_real / vc_std**2
np.testing.assert_allclose(classical_variogram(vc_true), vc_target_gamma, equal_nan=True)
print(f"{len(pairs):,} pairs, {valid_bins.sum()} valid bins, cutoff={max_dist:.3f} native coordinate units.")
```

## Constrained fit and candidate kernels

For each bandwidth, \(q_h=\gamma_Y(h)-v\), and
\[
b=\operatorname{clip}\left[\frac{\sum_h q_h(\gamma_U(h)-1)}{\sum_h q_h^2},0,1/v\right],\qquad a=1-bv.
\]
The fitting and plotting functions below are copied from the variance-constrained section of `variogram_matching.ipynb`. Adaptive `nn=0` remains the literal standardized permutation. The pure-noise and pure-smooth endpoints are permitted. Both theoretical and realized SSE are retained, with selection based on the theoretical objective.

```python
def variance_constrained_match(raw_fields, raw_variograms, labels, identity_index=None):
    """Fit a + b*gamma_Y subject to a + b*Var(Y) = 1, a,b >= 0."""
    smooth = (raw_fields - vc_mean) / vc_std
    smooth -= smooth.mean(axis=1, keepdims=True)
    variance = smooth.var(axis=1, ddof=0)
    gamma_y = raw_variograms / vc_std**2
    q = gamma_y[:, valid_bins, :] - variance[:, None, :]
    target_offset = vc_target_gamma[valid_bins] - 1.0
    denominator = np.sum(q**2, axis=1)
    numerator = np.sum(q * target_offset[None, :, :], axis=1)
    b = np.divide(numerator, denominator, out=np.zeros_like(numerator),
                  where=denominator > 0)
    upper = np.divide(1.0, variance, out=np.zeros_like(variance), where=variance > 0)
    b = np.clip(b, 0.0, upper)
    a = np.maximum(1.0 - b * variance, 0.0)
    if identity_index is not None:
        # Keep the same literal permutation baseline used earlier in the notebook.
        np.testing.assert_allclose(variance[identity_index], 1.0, atol=1e-12)
        a[identity_index] = 0.0
        b[identity_index] = 1.0
    fitted = a[:, None, :] + b[:, None, :] * gamma_y
    model_sse = np.sum((fitted[:, valid_bins] - vc_target_gamma[valid_bins])**2, axis=1)
    mixed = np.sqrt(b[:, None, :]) * smooth + np.sqrt(a[:, None, :]) * vc_noise
    if identity_index is not None:
        mixed[identity_index] = vc_permuted
    realized = np.stack([classical_variogram(field) for field in mixed])
    realized_sse = np.sum((realized[:, valid_bins] - vc_target_gamma[valid_bins])**2, axis=1)
    realized_variance = mixed.var(axis=1, ddof=0)
    cross_covariance = np.mean(smooth * vc_noise[None, :, :], axis=1)
    component_variance = a + b * variance
    np.testing.assert_allclose(component_variance, 1.0, atol=1e-12)
    np.testing.assert_allclose(realized_variance,
                               component_variance + 2 * np.sqrt(a * b) * cross_covariance,
                               atol=1e-12)
    best = model_sse.argmin(axis=0)
    genes = np.arange(vc_true.shape[1])
    best_field = np.column_stack([mixed[k, :, j] for j, k in enumerate(best)])
    best_gamma = np.column_stack([realized[k, :, j] for j, k in enumerate(best)])
    best_variance = realized_variance[best, genes]
    if np.any(best_variance <= 0):
        raise ValueError("A selected variance-constrained surrogate is constant.")
    # Diagnostic only: normalize the selected winner without changing its selection.
    normalized_gamma = best_gamma / best_variance
    normalized_sse = np.sum((normalized_gamma[valid_bins] - vc_target_gamma[valid_bins])**2, axis=0)
    summary = pd.DataFrame({
        "gene": selected_genes,
        "best_candidate": np.asarray(labels)[best],
        "a_noise": a[best, genes],
        "b_smooth": b[best, genes],
        "smooth_variance": variance[best, genes],
        "a_plus_b_v": component_variance[best, genes],
        "realized_variance": best_variance,
        "smooth_noise_covariance": cross_covariance[best, genes],
        "constrained_fit_SSE": model_sse[best, genes],
        "realized_SSE": realized_sse[best, genes],
        "SSE_if_final_normalized": normalized_sse,
    })
    return dict(labels=np.asarray(labels), a=a, b=b, smooth_variance=variance,
                fitted=fitted, realized=realized, model_sse=model_sse,
                realized_sse=realized_sse, realized_variance=realized_variance,
                best=best, best_field=best_field, best_gamma=best_gamma,
                normalized_gamma=normalized_gamma, summary=summary)


def plot_variance_constrained(result, family, candidate_label):
    colors = plt.colormaps["viridis"](np.linspace(0, 1, len(result["labels"])))
    fig, axes = plt.subplots(2, 5, figsize=(20, 8), constrained_layout=True)
    for j, (ax, gene) in enumerate(zip(axes.flat, selected_genes)):
        for k, label in enumerate(result["labels"]):
            ax.plot(lag_centers, result["realized"][k, :, j], color=colors[k],
                    linewidth=1.3, label=f"{candidate_label}={label:g}")
            ax.plot(lag_centers, result["fitted"][k, :, j], color=colors[k],
                    linestyle=":", linewidth=1.1,
                    label="Constrained theoretical fit" if k == 0 else None)
        ax.plot(lag_centers, vc_target_gamma[:, j], "o-", color="firebrick",
                markersize=4, linewidth=2, label="Real (unit variance)")
        ax.plot(lag_centers, vc_permuted_gamma[:, j], "s--", color="slategray",
                markersize=3, label="Permuted")
        best_label = result["labels"][result["best"][j]]
        ax.set(title=f"{gene} | best {candidate_label}={best_label:g}",
               xlabel="Distance (bin midpoint)", ylabel="Semivariance (standardized units)",
               xlim=(0, max_dist), ylim=(0, None))
        ax.grid(alpha=0.2)
    handles, labels = axes.flat[0].get_legend_handles_labels()
    fig.legend(handles, labels, loc="outside lower center", ncol=6, frameon=False)
    fig.suptitle(f"{family}: a + b Var(Y) = 1 — realized candidates, no final rescaling", fontsize=16)
    plt.show()

    fig, axes = plt.subplots(2, 5, figsize=(20, 8), constrained_layout=True)
    for j, (ax, gene) in enumerate(zip(axes.flat, selected_genes)):
        k = result["best"][j]
        ax.plot(lag_centers, vc_target_gamma[:, j], "o-", color="firebrick", label="Real")
        ax.plot(lag_centers, result["fitted"][k, :, j], ":", color="black", label="Constrained fit")
        ax.plot(lag_centers, result["best_gamma"][:, j], "^-", color="teal", label="Realized, unscaled")
        ax.plot(lag_centers, result["normalized_gamma"][:, j], "--", color="darkorange",
                label="If final normalized (diagnostic)")
        ax.set(title=gene, xlabel="Distance (bin midpoint)",
               ylabel="Semivariance (standardized units)", xlim=(0, max_dist), ylim=(0, None))
        ax.grid(alpha=0.2)
    handles, labels = axes.flat[0].get_legend_handles_labels()
    fig.legend(handles, labels, loc="outside lower center", ncol=4, frameon=False)
    fig.suptitle(f"{family}: winner selected by constrained theoretical SSE", fontsize=16)
    plt.show()
    display(result["summary"])
```

```python
LOCFIT_GAUSSIAN_FACTOR = 2.5
uniform_multipliers = np.array([1., 2., 3., 4., 6., 8., 12., 16., 24., 32., 48., 64.])
base_distance = float(np.median(cKDTree(coords).query(coords, k=2)[0][:, 1]))
uniform_bandwidths = base_distance * uniform_multipliers
nn_candidates = np.r_[0., .01, .02, .03, .04, .05, np.arange(1, 10, dtype=float) / 10]
distance2 = cdist(coords, coords, metric="sqeuclidean")
uniform_weights = []
adaptive_weights = []
with threadpool_limits(limits=1):
    for bandwidth in uniform_bandwidths:
        w = np.exp(-0.5 * LOCFIT_GAUSSIAN_FACTOR**2 * distance2 / bandwidth**2)
        uniform_weights.append(w / w.sum(axis=1, keepdims=True))
    for nn in nn_candidates[1:]:
        kth = int(np.ceil(nn * (n_cells - 1)))
        kth_distance2 = np.partition(distance2, kth, axis=1)[:, kth]
        if np.any(kth_distance2 <= 0):
            raise ValueError("Nonpositive adaptive bandwidth")
        w = np.exp(-0.5 * LOCFIT_GAUSSIAN_FACTOR**2 * distance2 / kth_distance2[:, None])
        adaptive_weights.append(w / w.sum(axis=1, keepdims=True))
print(f"Median nearest-neighbor distance d={base_distance:.3f}; "
      f"uniform bandwidth range={uniform_bandwidths[0]:.3f}–{uniform_bandwidths[-1]:.3f} native units.")
```

```python
results_by_seed = {}
summary_rows = []
candidate_rows = []
for seed in SEEDS:
    rng = np.random.default_rng(seed)
    perm = rng.permutation(n_cells)
    permuted_log_cpm = log_cpm[perm]
    vc_permuted = vc_true[perm]
    vc_permuted_gamma = classical_variogram(vc_permuted)
    with threadpool_limits(limits=1):
        vc_noise = (rng.standard_normal((n_cells, n_cells)) @ vc_true) / np.sqrt(n_cells)
        vc_noise -= vc_noise.mean(axis=0)
        vc_noise /= vc_noise.std(axis=0, ddof=0)
        uniform_raw = np.stack([w @ permuted_log_cpm for w in uniform_weights])
        adaptive_raw = np.stack([permuted_log_cpm] + [w @ permuted_log_cpm for w in adaptive_weights])
    uniform_gamma = np.stack([classical_variogram(y) for y in uniform_raw])
    adaptive_gamma = np.stack([classical_variogram(y) for y in adaptive_raw])
    uniform_result = variance_constrained_match(uniform_raw, uniform_gamma, uniform_multipliers)
    adaptive_result = variance_constrained_match(adaptive_raw, adaptive_gamma, nn_candidates, identity_index=0)
    results_by_seed[seed] = {"uniform": uniform_result, "adaptive": adaptive_result,
                             "permuted_gamma": vc_permuted_gamma.copy()}
    for family, result in [("uniform", uniform_result), ("adaptive", adaptive_result)]:
        table = result["summary"].copy()
        table.insert(0, "seed", seed)
        table.insert(1, "family", family)
        table["realized_RMSE"] = np.sqrt(table.realized_SSE / valid_bins.sum())
        table["constrained_RMSE"] = np.sqrt(table.constrained_fit_SSE / valid_bins.sum())
        table["at_upper_boundary"] = table.best_candidate == result["labels"][-1]
        table["permuted_RMSE"] = np.sqrt(np.mean((vc_permuted_gamma[valid_bins] - vc_target_gamma[valid_bins])**2, axis=0))
        if family == "uniform":
            old_indices = np.flatnonzero(uniform_multipliers <= 12)
            old_best = old_indices[result["model_sse"][old_indices].argmin(axis=0)]
            gene_indices = np.arange(n_genes)
            table["old_grid_best_d"] = uniform_multipliers[old_best]
            table["old_grid_realized_RMSE"] = np.sqrt(result["realized_sse"][old_best, gene_indices] / valid_bins.sum())
            table["old_grid_constrained_RMSE"] = np.sqrt(result["model_sse"][old_best, gene_indices] / valid_bins.sum())
        summary_rows.append(table)
        for k, label in enumerate(result["labels"]):
            candidate_rows.append(pd.DataFrame({"seed": seed, "family": family, "gene": selected_genes,
                "candidate": label, "constrained_SSE": result["model_sse"][k],
                "realized_SSE": result["realized_sse"][k], "realized_variance": result["realized_variance"][k]}))
    print(f"Finished seed {seed}")
summary = pd.concat(summary_rows, ignore_index=True)
candidate_scores = pd.concat(candidate_rows, ignore_index=True)
summary.to_csv(output_dir / "selected_fits.csv", index=False)
candidate_scores.to_csv(output_dir / "candidate_scores.csv", index=False)
display(summary.groupby("family").agg(
    gene_seed_fits=("gene", "size"), median_realized_RMSE=("realized_RMSE", "median"),
    maximum_realized_RMSE=("realized_RMSE", "max"), upper_boundary_count=("at_upper_boundary", "sum"),
    variance_min=("realized_variance", "min"), variance_max=("realized_variance", "max")))
display(summary[summary.family == "uniform"][["seed", "gene", "old_grid_best_d", "best_candidate",
    "old_grid_realized_RMSE", "realized_RMSE", "realized_variance"]])
```

```python
# Three-seed objective curves: solid theoretical SSE, dotted realized SSE.
fig, axes = plt.subplots(2, 5, figsize=(20, 8), constrained_layout=True)
for j, (ax, gene) in enumerate(zip(axes.flat, selected_genes)):
    for seed, color in zip(SEEDS, ["tab:blue", "tab:orange", "tab:green"]):
        result = results_by_seed[seed]["uniform"]
        ax.plot(uniform_multipliers, result["model_sse"][:, j], "o-", color=color,
                markersize=3, label=f"Seed {seed}: constrained")
        ax.plot(uniform_multipliers, result["realized_sse"][:, j], ":", color=color,
                label=f"Seed {seed}: realized")
    ax.axvline(12, color="black", linestyle="--", alpha=.5, label="Previous grid limit")
    ax.set(title=gene, xlabel="Bandwidth / median nearest-neighbor distance", ylabel="Variogram SSE",
           xscale="log", yscale="log")
    ax.grid(alpha=.2)
handles, labels = axes.flat[0].get_legend_handles_labels()
fig.legend(handles, labels, loc="outside lower center", ncol=4, frameon=False)
fig.suptitle("Maynard 151673 WM: expanded uniform bandwidth grid, three shared-draw comparisons", fontsize=15)
fig.savefig(output_dir / "uniform_bandwidth_sse.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
# Detailed variograms for the first seed; all three seeds are in the saved tables.
vc_permuted_gamma = results_by_seed[SEED]["permuted_gamma"]
plot_variance_constrained(results_by_seed[SEED]["uniform"], "Maynard 151673 WM — uniform Gaussian", "d")
plot_variance_constrained(results_by_seed[SEED]["adaptive"], "Maynard 151673 WM — adaptive Gaussian", "nn")
```

```python
# Compact old-grid versus expanded-grid comparison in standardized units.
fig, axes = plt.subplots(2, 5, figsize=(20, 8), constrained_layout=True)
result = results_by_seed[SEED]["uniform"]
old_indices = np.flatnonzero(uniform_multipliers <= 12)
old_best = old_indices[result["model_sse"][old_indices].argmin(axis=0)]
for j, (ax, gene) in enumerate(zip(axes.flat, selected_genes)):
    k = result["best"][j]
    ax.plot(lag_centers, vc_target_gamma[:, j], "o-", color="firebrick", label="Real")
    ax.plot(lag_centers, result["realized"][old_best[j], :, j], "--", color="slategray",
            label=f"Old grid: d={uniform_multipliers[old_best[j]]:g}")
    ax.plot(lag_centers, result["realized"][k, :, j], "^-", color="teal",
            label=f"Expanded: d={uniform_multipliers[k]:g}")
    ax.plot(lag_centers, result["fitted"][k, :, j], ":", color="black", label="Constrained fit")
    ax.set(title=gene, xlabel="Distance (native coordinate units)", ylabel="Semivariance (standardized)")
    ax.grid(alpha=.2)
    ax.legend(fontsize=7)
fig.suptitle("Maynard 151673 WM — seed 42: old versus expanded grid (same permutation and noise)", fontsize=15)
fig.savefig(output_dir / "uniform_old_vs_expanded.png", dpi=150, bbox_inches="tight")
plt.show()
```
